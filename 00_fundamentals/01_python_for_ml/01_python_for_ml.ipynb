## 1. Functions and data structures

ML code is mostly small functions that transform data. Keep each one short and give it one job.

```python
def rescale(values, low=0.0, high=1.0):
    """Map values linearly to [low, high]."""
    values = np.asarray(values, dtype=float)
    span = values.max() - values.min()
    return low + (values - values.min()) * (high - low) / span

print(rescale([3, 5, 9]))
config = {"seed": SEED, "test_size": 0.2, "model": "linear"}
print({k: v for k, v in config.items() if k != "model"})
```

## 2. Reproducibility basics

A result is only useful if someone can run it again. Fix the seed, record package versions and keep settings in one dictionary.

```python
from ml_lab.utils import print_environment
print_environment()
a = np.random.default_rng(SEED).normal(size=3)
b = np.random.default_rng(SEED).normal(size=3)
print("same seed, same numbers:", bool(np.allclose(a, b)))
```

## 3. Vectorised code is faster than loops

We time the same sum of squares with a Python loop and with NumPy on one million numbers.

```python
import time
x = np.random.default_rng(SEED).normal(size=1_000_000)
t = time.perf_counter(); s1 = sum(v * v for v in x); loop = time.perf_counter() - t
t = time.perf_counter(); s2 = float(np.dot(x, x)); vec = time.perf_counter() - t
print(f"loop {loop:.3f} s, numpy {vec:.4f} s, ratio {loop / vec:.0f}x, same result: {abs(s1 - s2) < 1e-6 * s2}")
```

## 4. Testing a small function

An `assert` is the smallest test. The repository runs real tests with `pytest` (see `tests/`).

```python
assert np.allclose(rescale([1, 2, 3]), [0, 0.5, 1])
assert rescale([10, 20], 0, 100).tolist() == [0.0, 100.0]
print("tests passed")
```
